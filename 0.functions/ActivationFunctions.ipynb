## Activation Functions in DeepLearning
An activation function is a small rule inside a neural network that decides how strongly a neuron should respond after it receives numbers. It takes the input number, changes it in a special way, and passes the result forward so the network can learn patterns instead of just doing simple addition.

```python
import matplotlib.pyplot as plt
inputx = np.arange(-5, 5, 0.01)
def plotfunc(x, title):
    plt.title(title)
    plt.plot(x)
    plt.axhline(y=0, color='r', linestyle='-')
    plt.show()
```

### 1. Identity. 
#### Most simple function. Output is same as input

```python
def identity(x): return x
plotfunc(
    identity(inputx),
    "Identity Function"
)
```

### 2. Binary step
#### Output is 1 only when input is not negative, else is 0

```python
def binary(xs): return [1 if x >= 0 else 0 for x in xs]
plotfunc(
    binary(inputx),
    "Binary Step Function"
)
```

### 3. Logistic, Sigmoid, Soft step
#### Divide the 1 by 1 adding negative exponential of negative of input 
<h3 style='color:orange'>0 <= output <= 1</h3>

```python
import numpy as np
def sigmoid(xs): return [1 / (1 + np.exp(-x)) for x in xs]
print("For negative: ", sigmoid([-0.7]))
print("For zero: ", sigmoid([0]))
print("For positive: ", sigmoid([0.7]))
plotfunc(
    sigmoid(inputx),
    "Sigmoid, Soft, Logistic Step Function"
)
```

### 4. Hyperbolic Tangent
#### Divide the sum of pos. and neg. exp.s of x / by deduction of  pos. and neg. exp.s of x
<h3 style='color:orange'>-1 <= output <= 1</h3>



```python
def tanh(xs): return [(np.exp(x)-np.exp(-x)) / (np.exp(x)+np.exp(-x)) for x in xs]
plotfunc(
    tanh(inputx),
    "Tanh Function"
)
```

### 4. ReLU
#### make all negatives zero
<h3 style='color:orange'>0 <= output <= infinity</h3>

```python
def relu(xs): return [0 if x < 0 else x for x in xs]
plotfunc(
    relu(inputx),
    "ReLU Function"
)
```
